```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline


loss= [10.0, 4.2, 3.5, 2.9, 2.5, 2.2, 2.0]
loss = torch.tensor(loss)
lossi = loss.log10()


plt.plot(loss)

```

```python
lossi = [5, 4, 3, 2, 1]

plt.plot(lossi)
plt.title("Training Loss")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.show()
```

```python
import numpy as np

grid = torch.randn(5,5)

print(grid)
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid.abs() > 0.5, cmap="gray")
plt.colorbar()
# you want to observe the white space not the black that is more than 0.5
#False → black
#True  → white
```

- imshow : image show or thing which plots grid for tensor
- abs() : absolute value means it just removes - sign from the numbers, eg -0.8 => 0.8
