```python
#things needed 
import torch
import torch.nn.functional as F
```

```python
# data preperation
xs, ys = [], []
for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        xs.append(ix1)
        ys.append(ix2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)
num = xs.nelement()
print('number of examples: ', num)

g = torch.Generator().manual_seed(2147483647)
W = torch.rand((27, 27), generator=g, requires_grad=True)
```

```python
#training loop
for k in range(100):
    xenc = F.one_hot(xs, num_classes=27).float()
    logits = xenc @ W
    counts = logits.exp()
    p = counts / counts.sum(1, keepdim=True)
    loss = -p[torch.arange(num), ys].log().mean()

    W.grad = None
    loss.backward()
    W.data += -2 * W.grad   # note: much larger LR than 0.01 — Karpathy uses -50

    if k % 10 == 0:
        print(k, loss.item())
```

###loss = -p[torch.arange(num), ys].log().mean():
- here we are calculating the loss where torch.arange(num) and ys are both same size tensors and PyTorch is optimized to take both tensor and single number as indexing, which return same size of vector, if not understood read below 2 points and test it with word emma at the bottom of the code 

- Indexing a tensor with tensor[index_tensor_1, index_tensor_2], where both index tensors have the same length N, always produces N results — one per position — by zipping the two index tensors together as coordinate pairs. This is fundamentally different from indexing with a single integer (p[2], grabs one whole row) or a slice (p[1:3], grabs a contiguous chunk). Two same-length index tensors = "grab these specific (row, col) pairs."

- torch.arange(num) supplies the row coordinate (which example), ys supplies the column coordinate (which class was correct for that example) — and because both are tensors of the same length (num), PyTorch does this one-shot parallel gather instead of forcing you to write a Python loop. That's the entire reason this line is fast even across 228,146 real training examples — it's one vectorized C++ operation instead of 228,146 Python loop iterations.

```python
g = torch.Generator().manual_seed(2147483647)

for i in range(5):

    out = []
    
    ix =0 
    while True:

        xenc = F.one_hot(torch.tensor([ix]), num_classes=27).float()
        logits = xenc @ W
        counts = logits.exp()
        p = counts / counts.sum(1, keepdim=True)
        
        ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
        out.append(itos[ix])
        if ix == 0:
            break
    print(''.join(out))


```
