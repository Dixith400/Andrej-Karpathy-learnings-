# Forward Pass — Variable Abbreviations

| Variable | Full Form | Meaning |
|---|---|---|
| `Xb` | **Input Batch** | A mini-batch of input examples containing character/token indices. |
| `Yb` | **Target Batch** | The correct target character/token for each example in `Xb`. |
| `C` | **Character Embedding Matrix** | Matrix that stores the embedding vector for every character/token. |
| `emb` | **Embedding** | Embedding vectors selected from `C` using the indices in `Xb`. |
| `embcat` | **Embedding Concatenated** | The embedding vectors flattened/concatenated into one vector per example. |
| `W1` | **Weight Matrix 1** | Weights of the first linear layer. |
| `b1` | **Bias 1** | Bias of the first linear layer. |
| `hprebn` | **Hidden Pre-Batch-Normalization** | Activations produced by the first linear layer before Batch Normalization. |
| `bnmeani` | **Batch Normalization Mean** | Mean of `hprebn` across the batch for each hidden neuron. |
| `bndiff` | **Batch Normalization Difference** | Difference between each activation and its batch mean. |
| `bndiff2` | **Batch Normalization Difference Squared** | Squared difference from the batch mean. |
| `bnvar` | **Batch Normalization Variance** | Variance of the activations within the batch. |
| `bnvar_inv` | **Batch Normalization Variance Inverse** | Inverse square root of the variance, used to normalize the activations. |
| `bnraw` | **Batch Normalization Raw Output** | Normalized activations before applying the learnable scale and shift. |
| `bngain` | **Batch Normalization Gain** | Learnable scaling parameter, often called `gamma` in BatchNorm. |
| `bnbias` | **Batch Normalization Bias** | Learnable shifting parameter, often called `beta` in BatchNorm. |
| `hpreact` | **Hidden Pre-Activation** | Batch-normalized activation after applying gain and bias, immediately before `tanh`. |
| `h` | **Hidden Activation** | Output of the `tanh` non-linearity. |
| `W2` | **Weight Matrix 2** | Weights of the second linear/output layer. |
| `b2` | **Bias 2** | Bias of the second linear/output layer. |
| `logits` | **Logits** | Raw, unnormalized scores produced by the output layer for each possible class. |
| `logit_maxes` | **Logit Maximums** | Maximum logit in each example, used for numerical stability. |
| `norm_logits` | **Normalized Logits** | Logits after subtracting the maximum logit from each example. |
| `counts` | **Exponentiated Logits / Counts** | Exponentials of the normalized logits. |
| `counts_sum` | **Sum of Counts** | Sum of the exponentiated logits for each example. |
| `counts_sum_inv` | **Inverse of Sum of Counts** | Reciprocal of `counts_sum`, used to normalize the counts into probabilities. |
| `probs` | **Probabilities** | Predicted probability distribution over all possible classes. |
| `logprobs` | **Log Probabilities** | Natural logarithm of the predicted probabilities. |
| `loss` | **Loss** | Average negative log probability assigned to the correct target classes. |
| `n` | **Number of Examples** | Number of examples in the current mini-batch. |
